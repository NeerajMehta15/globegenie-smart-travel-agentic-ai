```python
from agent.input_analyzer import InputAnalyzer
from agent.orchestrator import Orchestrator

# Test input
user_input = "5 day trip to North Vietnam, budget $3000"

# Run the workflow
analyzer = InputAnalyzer()
trip_state = analyzer.process_input(user_input)

orchestrator = Orchestrator()
final_state = orchestrator.orchestrate_trip_planning(trip_state)
```

```python
# CHECK 1: What is final_state?
print("="*50)
print("1. FINAL STATE TYPE:")
print("Type:", type(final_state))
print("Is dict?", isinstance(final_state, dict))
print()
```

```python
# CHECK 2: What's inside?
print("="*50)
print("2. WHAT'S INSIDE FINAL_STATE:")
if isinstance(final_state, dict):
    print("Keys:", list(final_state.keys()))
    print()
    
    # Check each key
    for key in final_state.keys():
        print(f"  {key}: {type(final_state[key])}")
else:
    print("Attributes:", [attr for attr in dir(final_state) if not attr.startswith('_')])
print()
```

```python
 CHECK 3: How to access final_plan?
print("="*50)
print("3. ACCESSING FINAL_PLAN:")
try:
    if isinstance(final_state, dict):
        final_plan = final_state['final_plan']
        print("✅ Dict access works: final_state['final_plan']")
    else:
        final_plan = final_state.final_plan
        print("✅ Attribute access works: final_state.final_plan")
    
    print("Final plan type:", type(final_plan))
    
    if isinstance(final_plan, dict):
        print("Final plan keys:", list(final_plan.keys()))
    elif isinstance(final_plan, str):
        print("⚠️ FINAL PLAN IS A STRING!")
        print("First 200 chars:", final_plan[:200])
except Exception as e:
    print(f"❌ Error: {e}")
print()

# CHECK 4: What's in itinerary?
print("="*50)
print("4. CHECKING ITINERARY:")
try:
    if isinstance(final_plan, dict):
        itinerary = final_plan.get('itinerary', [])
        print("Itinerary type:", type(itinerary))
        print("Number of items:", len(itinerary) if itinerary else 0)
        if itinerary:
            print("First item:", itinerary[0])
    else:
        print("Final plan is not a dict, can't check itinerary")
except Exception as e:
    print(f"❌ Error: {e}")
print()

# CHECK 5: What's in destination_research?
print("="*50)
print("5. CHECKING DESTINATION RESEARCH:")
try:
    if isinstance(final_state, dict):
        dest_research = final_state.get('destination_research')
    else:
        dest_research = getattr(final_state, 'destination_research', None)
    
    print("Research type:", type(dest_research))
    if isinstance(dest_research, dict):
        print("Research keys:", list(dest_research.keys()))
        print("Content:", dest_research)
except Exception as e:
    print(f"❌ Error: {e}")
```
